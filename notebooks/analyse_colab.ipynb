# Analyse SUJ1 — Tobii
Exécuter les cellules dans Google Colab. Transférer uniquement des données autorisées par Airbus.
Les données originales restent hors de GitHub. Le script est spécifique au schéma SUJ1.

```python
from pathlib import Path
import subprocess, sys
if not Path("AirbusHelico/.git").exists():
    subprocess.run(["git", "clone", "https://github.com/emmachabani/AirbusHelico.git"], check=True)
subprocess.run([sys.executable, "-m", "pip", "install", "-r", "AirbusHelico/requirements.txt"], check=True)
```

```python
from google.colab import files
uploaded = files.upload()
if len(uploaded) != 1:
    raise ValueError("Choisir uniquement SUJ1.xlsx")
source = next(iter(uploaded))
subprocess.run([sys.executable, "AirbusHelico/scripts/analyse_tobii.py", source, "--output", "resultats_SUJ1"], check=True)
```

```python
from IPython.display import display, Image
import shutil
for image in sorted(Path("resultats_SUJ1").glob("*.png")):
    display(Image(filename=str(image)))
shutil.make_archive("resultats_SUJ1", "zip", "resultats_SUJ1")
files.download("resultats_SUJ1.zip")
```
