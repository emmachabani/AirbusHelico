```python
from pathlib import Path
import subprocess, sys
if not Path("AirbusHelico/.git").exists():
    subprocess.run(["git", "clone", "https://github.com/emmachabani/AirbusHelico.git"], check=True)
subprocess.run([sys.executable, "-m", "pip", "install", "-r", "AirbusHelico/requirements.txt"], check=True)
```

```python
from google.colab import files
subject = "SUJ1"  # choisir SUJ1, SUJ2, SUJ3, SUJ5, SUJ6 ou SUJ7 selon le fichier
uploaded = files.upload()
if len(uploaded) != 1:
    raise ValueError("Choisir un seul fichier Excel")
source = next(iter(uploaded))
output = "resultats_" + subject
subprocess.run([sys.executable, "AirbusHelico/scripts/analyse_tobii.py", source, "--subject", subject, "--output", output], check=True)
```

```python
from IPython.display import display, Image
import shutil
for image in sorted(Path(output).glob("*.png")):
    display(Image(filename=str(image)))
shutil.make_archive(output, "zip", output)
files.download(output + ".zip")
```
